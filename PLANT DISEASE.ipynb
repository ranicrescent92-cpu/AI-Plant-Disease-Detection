```python
from pathlib import Path

data_dir = Path("data/raw/plantvillage dataset/color")

classes = [folder.name for folder in data_dir.iterdir() if folder.is_dir()]

print("Number of classes:", len(classes))

for cls in sorted(classes):
    print(cls)
```

```python
from collections import OrderedDict

class_counts = OrderedDict()

for cls in sorted(classes):
    class_path = data_dir / cls

    count = len([
        file for file in class_path.iterdir()
        if file.suffix.lower() in [".jpg", ".jpeg", ".png"]
    ])

    class_counts[cls] = count

for cls, count in class_counts.items():
    print(f"{cls}: {count}")
```

```python
total_images = sum(class_counts.values())

print("Total images:", total_images)
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import random

sample_classes = random.sample(sorted(classes), 6)

plt.figure(figsize=(12, 8))

for i, cls in enumerate(sample_classes):
    class_path = data_dir / cls

    image_files = [
        file for file in class_path.iterdir()
        if file.suffix.lower() in [".jpg", ".jpeg", ".png"]
    ]

    image_path = random.choice(image_files)

    image = Image.open(image_path)

    plt.subplot(2, 3, i + 1)
    plt.imshow(image)
    plt.title(cls)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
